# **Notação Big-O**

```python
from math import log

import matplotlib.pyplot as plt
import numpy as np
```

## Função 1 - O(n)

```python
# 11 passos
def soma1(n):
    soma = 0
    for i in range(n + 1):
        soma += i
    return soma
```

```python
soma1(10)
```

```python
%timeit soma1(10)
```

## Função 2

```python
# 3 passos 0(3)
def soma2(n):
    return (n * (n + 1)) /2
```

```python
soma2(10)
```

```python
%timeit soma2(10)
```

## Função 3

```python
def lista1():
    lista = []
    for i in range(1000):
        lista += [i]
    return lista
```

```python
print(lista1())
```

```python
len(lista1())
```

```python
%timeit lista1()
```

```python
def lista12():
    lista = []
    for i in range(1000):
        lista.append(i)  # noqa: PERF402
    return lista
```

## Funções Big-O
- 1 Constant
- log(n) Logarithmic
- n Linear
- nlog(n) Log Linear
- n^2 Quadratic
- n^3 Cubic
- 2^n Exponential

```python
n = np.linspace(1, 10, 100)
```

```python
len(n)
```

```python
n
```

```python
labels = [
    'Constant', 'Logarithmic', 'Linear',
    'Log Linear', 'Quadratic', 'Cubic', 'Exponential'
]
big_o = [
    np.ones(n.shape), np.log(n), n, n * np.log(n), n**2, n**3, 2**n
]
```

```python
plt.figure(figsize=(10,8))
plt.ylim(0, 100)
for i in range(len(big_o)):
    plt.plot(n, big_o[i], label=labels[i])
plt.legend()
plt.ylabel('Runtime')
plt.xlabel('n');
```

## Exemplos Big-O

## Constant - O(1)

```python
lista = [1, 2, 3, 4, 5]
```

```python
def constant(n):
    print(n[0])
```

```python
constant(lista)
```

## Linear - O(n)

```python
def linear(n):
    for i in n:
        print(i)
```

```python
linear(lista)
```

## Quadratic O(n²) - Polynomial

```python
def quadratic(n):
    for i in n:
        for j in n:
            print(i, j)
        print('-'*5)
```

```python
quadratic(lista)
```

## Combination

```python
def combination(n): # O(n)
    # O(1)
    print(n[0])
    # O(5)
    for i in range(5):
        print(f'test {i}')
    # O(n)
    for i in n:
        print(i)
    # O(n)
    for i in n:
        print(i)
    # O(3)
    print('Python')
    print('Python')
    print('Python')
```

```python
combination(lista)
```
