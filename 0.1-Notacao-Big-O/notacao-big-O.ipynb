# **Notação Big-O**

## Função 1 - O(n)

```python
# 11 passos
def soma1(n):
    soma = 0
    for i in range(n + 1):
        soma += i
    return soma
```

```python
soma1(10)
```

```python
%timeit soma1(10)
```

## Função 2

```python
# 3 passos 0(3)
def soma2(n):
    return (n * (n + 1)) /2
```

```python
soma2(10)
```

```python
%timeit soma2(10)
```
